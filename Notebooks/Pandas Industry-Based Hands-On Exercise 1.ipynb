## Challenge: Customer Value Classification

A new variable, `Customer_Value`, will be created to classify customer records according to their sales value. Records with sales of ₦100,000 or more will be classified as High Value, while records with sales below ₦100,000 will be classified as Standard.

```python
df["Customer_Value"] = df["Sales"].apply(
    lambda x: "High Value" if x >+ 100000 else "Standard"
)
```

```python
df.head()
```

### Interpretation

The `Customer_Value` column was successfully added to the dataset. The classification rule correctly assigns records with sales values below ₦100,000 to the Standard category. The first five records all have sales values below the ₦100,000 threshold and are therefore classified as Standard.

```python
df["Customer_Value"].value_counts()
```

### Customer Value Classification

The customer-value classification shows that 2,983 records are classified as High Value, while 2,017 records are classified as Standard. This indicates that High Value records constitute the larger proportion of the dataset. Specifically, approximately 59.66% of the records meet or exceed the ₦100,000 sales threshold, while 40.34% have sales values below the threshold.
