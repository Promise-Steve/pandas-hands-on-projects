## Data Cleaning

Based on the data-quality assessment, the dataset contains duplicate records, inconsistent formats in the Visit_Count variable, and missing values. The cleaning process will address these issues systematically while preserving valid patient information.

```python
df = df.drop_duplicates() ### --- dropping duplicates
```

```python
df.duplicated().sum() ### --- verifying if duplicates still remain
```

```python
df = df.copy()
```

```python
df["Visit_Count"] = df["Visit_Count"].replace({
    "six": "6",
    "seven": "7",
    "ten": "10",
    "4 visits": "4"
})
```

```python
df["Visit_Count"]                        
```

```python
df["Visit_Count"].unique()
```

```python
df["Visit_Count"] = pd.to_numeric(df["Visit_Count"], errors = "coerce")
```

```python
df.info()
```

### Standardisation of Visit Count

The inconsistent entries in the `Visit_Count` variable were standardised by converting textual representations such as `six`, `seven`, `ten`, and `4 visits` to their corresponding numerical values. The variable was subsequently converted from an `object` data type to a numerical (`float64`) data type. After removing four duplicate records, the cleaned dataset contains 2,996 records, with 2,955 non-null observations for Visit_Count.

```python
df["Age"].mean(), df["Age"].median()
```

### Treatment of Missing Age Values

The mean and median ages in the cleaned dataset are approximately 49.00 and 49.00 years, respectively. Because the two measures are essentially identical, the median age of 49 years will be used to replace the missing Age values. This provides a simple central-value treatment while preserving the existing patient records.

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Age"].isnull().sum()
```

### Age Cleaning Result

The missing Age values were replaced using the median age of 49 years. A subsequent missing-value check returned zero, confirming that all missing Age observations have been successfully addressed.

```python
df["City"].value_counts()
```

### Investigation of City Values

The City variable contains several inconsistencies in how city names are recorded. For example, Abuja appears as both `Abuja` and `ABUJA`, Ibadan appears as both `Ibadan` and `ibadan`, Lagos appears as both `Lagos` and `lagos`, and Port Harcourt appears as both `Port Harcourt` and `Port-Harcourt`. In addition, five records contain `TEST`, which does not represent a valid city entry in the dataset. These inconsistencies should be standardised before the missing City values are addressed.

```python
df[df["City"] == "TEST"]
```

### Investigation of Test Records

Further inspection of the five records with `TEST` as the City shows that they are test or dummy records rather than genuine patient records. The records contain indicators such as `System Test`, `Test Patient One`, `Dummy Patient`, `Clinic Test Entry`, and `Sample Record`, with `Unknown` gender and `Test` visit types. These records are therefore considered invalid patient records and should be removed from the analytical dataset before further analysis.

```python
df = df[df["City"] != "TEST"].copy()
```

```python
df["City"].value_counts()
```

```python
df["City"] = df["City"].replace({
    "ABUJA" : "Abuja",
    "ibadan": "Ibadan",
    "lagos" : "Lagos",
    "Port-Harcourt": "Port Harcourt"
})
```

```python
df["City"].value_counts()
```

```python
df.loc[df["City"].str.lower().str.contains("ibadan",na=False), "City"].unique()
```

```python
df["City"] = df["City"].str.strip()
```

```python
df["City"].value_counts()
```

```python
df["City"] = df["City"].replace({
    "ibadan": "Ibadan"
})
```

```python
df["City"].value_counts()
```

### Standardisation of City Names

Inconsistent city entries were standardised to ensure that different spellings and formatting variations represented the same city. Leading and trailing whitespace was removed, and inconsistent entries such as `ABUJA`, `ibadan`, `lagos`, and `Port-Harcourt` were standardised to `Abuja`, `Ibadan`, `Lagos`, and `Port Harcourt`, respectively. The invalid `TEST` records were previously removed because they represented test or dummy patient records.

### Final City Standardisation

After removing invalid test records and standardising inconsistent city names, the City variable contains 18 valid city categories. Variations in capitalisation, punctuation, and trailing whitespace were corrected so that entries referring to the same city are represented consistently. This standardisation improves the accuracy of subsequent frequency analyses of patient locations.

```python
df["City"].isnull().sum()
```

### Treatment of Missing City Values

After removing invalid test records and standardising the valid city names, 118 patient records still have missing City values. Because City is a categorical variable and there is insufficient information in the affected records to determine the patients' actual locations, the missing values will not be arbitrarily assigned to the most frequent city. ### Treatment of Missing Visit Count Values

The cleaned `Visit_Count` variable has a mean of approximately 6.52 visits and a median of 7 visits, with values ranging from 1 to 12 visits. Because the mean and median are relatively close, the median value of 7 visits will be used to replace the remaining missing Visit_Count values. This preserves a realistic whole-number value for the count variable.

```python
df["Visit_Count"] = df["Visit_Count"].fillna(df["Visit_Count"].median())
```
